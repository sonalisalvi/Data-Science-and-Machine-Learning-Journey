## Creating Student Performance Dataset

We create a small dataset containing:

- Student Name
- Age
- Marks
- City

This dataset will help us understand different types of data.

```python
students = pd.DataFrame({
    "Name": [
        "John",
        "Erik",
        "Sara",
        "Oskar",
        "Jane"
     ],

     "Age": [
        20,
        21,
        19,
        22,
        23
    ],

    "Marks": [
        85,
        90,
        60,
        95,
        70
    ],

    "City": [
        "Mumbai",
        "Delhi",
        "Pune",
        "Mumbai",
        "Delhi"
    ]
})

print(students)

students
```

## Understanding Dataset Shape

Rows represent observations.

Columns represent variables/features.

```python
numerical_columns = students.select_dtypes(
    include = np.number
)

numerical_columns
```
